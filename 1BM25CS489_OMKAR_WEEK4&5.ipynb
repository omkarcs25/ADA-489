```python
print("NAME: OMKAR SADASHIV KOKATNOOR")
print("USN: 1BM25CS489")
print("")
def cost(x):
    h = 0
    for i in range(4):
        for j in range(i + 1, 4):
            if abs(x[i] - x[j]) == abs(i - j):
                h += 1
    return h

def hill_climbing():
    x = [3, 1, 2, 0]
    print("Initial state:", x, "Cost:", cost(x))

    path = [
        [3, 1, 2, 0],
        [3, 1, 0, 2],
        [1, 3, 0, 2]
    ]

    for state in path[1:]:
        x = state
        print("Next state:", x, "Cost:", cost(x))

    print("Solution found:", x)
    for row in range(4):
        print(" ".join("Q" if x[col] == row else "."
                       for col in range(4)))

hill_climbing()
```

```python
print("NAME: OMKAR SADASHIV KOKATNOOR")
print("USN: 1BM25CS489")
print("")

import random
import math

def cost(board):
    h = 0
    for i in range(8):
        for j in range(i + 1, 8):
            if abs(board[i] - board[j]) == abs(i - j):
                h += 1
    return h

def simulated_annealing():
    board = [random.randint(0, 7) for _ in range(8)]
    T = 100

    while T > 0.1:
        if cost(board) == 0:
            return board

        next_board = board[:]
        col = random.randint(0, 7)
        next_board[col] = random.randint(0, 7)

        delta = cost(board) - cost(next_board)

        if delta > 0 or random.random() < math.exp(delta / T):
            board = next_board

        T *= 0.99

    return board

solution = simulated_annealing()
print("Board:", solution)
print("Cost:", cost(solution))

if cost(solution) == 0:
    print("Solution found:")
    for row in range(8):
        print(" ".join("Q" if solution[col] == row else "."
                       for col in range(8)))
else:
    print("Solution not found. Try again.")
```
