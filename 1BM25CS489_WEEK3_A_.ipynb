```python
import heapq

print("NAME: OMKAR SADASHIV KOKATNOOR")
print("USN: 1BM25CS489")
class PuzzleState:
    def __init__(self, board, g=0, parent=None, move=""):
        self.board = tuple(board)
        self.g = g
        self.parent = parent
        self.move = move
        self.h = self.calculate_manhattan_distance()
        self.f = self.g + self.h

    def calculate_manhattan_distance(self):
        # Goal positions map
        goal_positions = {
            1: (0, 0), 2: (0, 1), 3: (0, 2),
            8: (1, 0), 'gap': (1, 1), 4: (1, 2),
            7: (2, 0), 6: (2, 1), 5: (2, 2)
        }

        distance = 0
        for i, val in enumerate(self.board):
            if val == 'gap':
                continue
            curr_row, curr_col = i // 3, i % 3
            goal_row, goal_col = goal_positions[val]
            distance += abs(curr_row - goal_row) + abs(curr_col - goal_col)
        return distance

    def get_neighbors(self):
        neighbors = []
        gap_idx = self.board.index('gap')
        row, col = gap_idx // 3, gap_idx % 3

        # Define possible moves for the gap: (d_row, d_col, direction_name)
        moves = [
            (-1, 0, 'Up'),
            (1, 0, 'Down'),
            (0, -1, 'Left'),
            (0, 1, 'Right')
        ]

        for dr, dc, move_name in moves:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_gap_idx = new_row * 3 + new_col
                new_board = list(self.board)
                # Swap gap with the target tile
                new_board[gap_idx], new_board[new_gap_idx] = new_board[new_gap_idx], new_board[gap_idx]
                neighbors.append(PuzzleState(new_board, self.g + 1, self, move_name))
        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def reconstruct_path(state):
    path = []
    curr = state
    while curr:
        path.append(curr)
        curr = curr.parent
    return path[::-1]

def print_board(board):
    for i in range(0, 9, 3):
        print(f"[ {board[i]:^5} | {board[i+1]:^5} | {board[i+2]:^5} ]")
    print("-" * 27)

def solve_a_star(initial_board, goal_board):
    initial_state = PuzzleState(initial_board)
    goal_tuple = tuple(goal_board)

    open_set = []
    heapq.heappush(open_set, initial_state)
    closed_set = set()

    nodes_explored = 0

    while open_set:
        current_state = heapq.heappop(open_set)
        nodes_explored += 1

        if current_state.board == goal_tuple:
            return reconstruct_path(current_state), nodes_explored

        closed_set.add(current_state.board)

        for neighbor in current_state.get_neighbors():
            if neighbor.board in closed_set:
                continue
            heapq.heappush(open_set, neighbor)

    return None, nodes_explored

# Define States
initial_state = [2, 8, 3, 1, 6, 4, 7, 'gap', 5]
goal_state = [1, 2, 3, 8, 'gap', 4, 7, 6, 5]

# Run Algorithm
path, total_explored = solve_a_star(initial_state, goal_state)

if path:
    print(f"Successfully found a solution in {len(path) - 1} steps!")
    print(f"Total nodes explored: {total_explored}\n")
    for step, state in enumerate(path):
        if step == 0:
            print("Initial State:")
        else:
            print(f"Step {step}: Move gap {state.move}")
        print_board(state.board)
else:
    print("No solution found.")
```
